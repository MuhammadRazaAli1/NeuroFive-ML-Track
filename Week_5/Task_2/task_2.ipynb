```python
# Import necessary libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import joblib
import warnings
warnings.filterwarnings('ignore')

# 1. Load the dataset
# Make sure the path matches where your train.csv is located
df = pd.read_csv('../dataset/train.csv')

# 2. Simple Data Cleaning & Encoding
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
# Encode Sex: Male = 0, Female = 1
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# 3. Select Features & Target
features = ['Pclass', 'Sex', 'Age', 'Fare']
X = df[features]
y = df['Survived']

# 4. Train the Random Forest Model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

# 5. Save the model to a file using joblib
joblib.dump(model, 'titanic_model.joblib')
print("Model successfully saved as 'titanic_model.joblib'")
```
